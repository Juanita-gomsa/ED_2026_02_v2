```python
def find_min(T:LinkedBinaryTree): 
    #Retorna el contenedor donde se encuentra el valor minimo
    current = T.root()
    while T.left(current) != None:
        current = T.left(current)
    return current 
find_min(l).element()
```

```python
def es_ordenado(T:LinkedBinaryTree):
    #Revisar si el arbol esta ordenado
    i = T.inorder()
    a = next(i)
    for j in range(len(T)-2):
        b= next(i)
        if b.element < a.element():
            return False
        a = next(i)
        a, b = b, a 
    return True 
        
```
