## 1. Generación de tus listas

Escribe tu número de documento de identidad (solo dígitos, sin puntos ni espacios) en
`CEDULA` y ejecuta la celda. Esto fija la semilla de `random` y genera dos listas
enlazadas propias, `L1` y `L2`, con tamaños y valores distintos para cada estudiante.

```python
from goodrich.exceptions import Empty


class SinglyLinkedList:
    """Lista enlazada simple con referencias a head y tail."""

    class _Node:
        """Nodo interno: guarda el elemento y la referencia al siguiente."""
        __slots__ = '_element', '_next'
        def __init__(self, element, nxt):
            self._element = element
            self._next    = nxt

    def __init__(self):
        self._head = None
        self._tail = None
        self._size = 0

    def __len__(self):    return self._size
    def is_empty(self):   return self._size == 0

    def first(self):
        if self.is_empty():
            raise Empty('lista vacía')
        return self._head._element

    def last(self):
        if self.is_empty():
            raise Empty('lista vacía')
        return self._tail._element

    def insert_first(self, e):
        self._head = self._Node(e, self._head)
        if self._size == 0:
            self._tail = self._head
        self._size += 1

    def insert_last(self, e):
        node = self._Node(e, None)
        if self.is_empty():
            self._head = node
        else:
            self._tail._next = node
        self._tail = node
        self._size += 1

    def remove_first(self):
        if self.is_empty():
            raise Empty('lista vacía')
        val = self._head._element
        self._head = self._head._next
        self._size -= 1
        if self._size == 0:
            self._tail = None
        return val


# --- Prueba rápida ---
S = SinglyLinkedList()
for x in ['A', 'B', 'C', 'D']:
    S.insert_last(x)
print(len(S), S.first(), S.last())   # 4 A D
S.remove_first()
print(len(S), S.first())             # 3 B
```

## 2. Tu solución

Implementa `resolver(L1, L2)`:

1. Concatena `L2` al final de `L1`
2. Retorna el **elemento** del penúltimo nodo de la lista `L1` ya concatenada (el nodo
   justo antes del último).

```python
def resolver(L1, L2):
    if L2.is_empty():
        pass
    elif L1.is_empty():
        L1._head = L2._head
        L1._tail = L2._tail
        L1._size = L2._size
    else:
        L1._tail._next = L2._head
        L1._tail = L2._tail
        L1._size = L1._size + L2._size

    L2._head = None
    L2._tail = None
    L2._size = 0

    ac = L1._head
    nodo_anterior = None

    while ac._next is not None:
        nodo_anterior = ac
        ac = ac._next

    return nodo_anterior._element


resultado = resolver(L1, L2)
print(resultado)
```
