```python
class cola_prioridad():
    def __init__(self):
        self.elementos = []
    def __len__(self):
        return len(self.elementos)
    
    def add(self, k, v ):
        self.elementos.append((k, v))
    
    def extraer_minimo(self):
        if not self.elementos:
            return None
        minimo = min(self.elementos, key=lambda x: x[0])
        self.elementos.remove(minimo)
        return minimo
    # Tarea hacer la version ordenada para es decir de mnaera menor a mayor y que extraiga el minimo
    
    def is_empty(self):
        return len(self.elementos) == 0
```

```python
cp = cola_prioridad()
for i in [2, -5, 9, 4, 3]:
    cp.add(i, i**2)
```

```python
class Queue:
    def __init__(self):
        self._priority_queue = UnsortedPriorityQueue()
        self._next_priority = 0

    def __len__(self):
        return len(self._priority_queue)

    def is_empty(self):
        return self._priority_queue.is_empty()

    def enqueue(self, value):
        self._priority_queue.add(self._next_priority, value)
        self._next_priority += 1

    def dequeue(self):
        _, value = self._priority_queue.remove_min()
        return value

    def first(self):
        _, value = self._priority_queue.min()
        return value


class Stack:

    def __init__(self):
        self._data = UnsortedPriorityQueue()
        self._next = 0

    def __len__(self):
        return len(self._data)

    def is_empty(self):
        return self._data.is_empty()

    def push(self, value):
        self._data.add(self._next, value) #Hace que asigne prioridades de manera decreciente, es decir, el último elemento agregado tendrá la prioridad más baja, y por lo tanto será el primero en ser extraído.
        self._next  -= 1

    def pop(self):
        if self.is_empty():
            raise Exception("Stack is empty")
        return self._data.remove_min()[1]

    def top(self):#Retorna elultimo elemento agregado a la pila sin eliminarlo, es decir, el elemento con la prioridad más baja.                                                 
       return self._data.min()[1]
```
